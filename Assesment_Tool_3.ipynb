```python
n, k = map(int, input().split())
alphabet = input().split()
start = int(input())

x = list(map(int, input().split()))
final = set(x[1:])

t = int(input())
trans = {}

for i in range(t):
    p, c, r = input().split()
    p = int(p)
    r = int(r)
    trans.setdefault((p, c), []).append(r)

q = int(input())

for i in range(q):
    s = input()
    if s == "#":
        s = ""

    states = {start}

    for c in s:
        new = set()
        for p in states:
            for r in trans.get((p, c), []):
                new.add(r)
        states = new

    if states & final:
        print("ACCEPT")
    else:
        print("REJECT")
```

```python
vowels = "aeiou"

def double(w):
    if len(w) < 3 or w[-1] in "wxy":
        return False
    if w[-3] in vowels or w[-2] not in vowels or w[-1] in vowels:
        return False

    groups = 0
    inside = False

    for c in w:
        if c in vowels:
            if not inside:
                groups += 1
            inside = True
        else:
            inside = False

    return groups == 1

def forms(w, pos):
    a = []

    if pos in ("N", "V"):
        if w.endswith(("s", "x", "z", "ch", "sh")):
            a.append((w + "es", "PL" if pos == "N" else "3SG"))
        if len(w) >= 2 and w[-1] == "o" and w[-2] not in vowels:
            a.append((w + "es", "PL" if pos == "N" else "3SG"))
        if w.endswith("y") and w[-2] not in vowels:
            a.append((w[:-1] + "ies", "PL" if pos == "N" else "3SG"))
        a.append((w + "s", "PL" if pos == "N" else "3SG"))

    if pos == "V":
        if len(w) >= 3 and w.endswith("e") and w[-2] not in vowels:
            a.append((w[:-1] + "ing", "PROG"))
        elif double(w):
            a.append((w + w[-1] + "ing", "PROG"))
        else:
            a.append((w + "ing", "PROG"))

        if w.endswith("e"):
            a.append((w + "d", "PAST"))
        elif w.endswith("y") and w[-2] not in vowels:
            a.append((w[:-1] + "ied", "PAST"))
        elif double(w):
            a.append((w + w[-1] + "ed", "PAST"))
        else:
            a.append((w + "ed", "PAST"))

    if pos == "ADJ":
        if w.endswith("e"):
            a.append((w + "r", "COMP"))
            a.append((w + "st", "SUP"))
        elif w.endswith("y") and w[-2] not in vowels:
            a.append((w[:-1] + "ier", "COMP"))
            a.append((w[:-1] + "iest", "SUP"))
        elif double(w):
            a.append((w + w[-1] + "er", "COMP"))
            a.append((w + w[-1] + "est", "SUP"))
        else:
            a.append((w + "er", "COMP"))
            a.append((w + "est", "SUP"))

    return a

S = int(input())
lex = []

for i in range(S):
    w, p = input().split()
    lex.append((w, p))

Q = int(input())

for i in range(Q):
    word = input()
    ans = []

    for stem, pos in lex:
        if word == stem:
            ans.append(stem + "+" + pos)

        for form, feature in forms(stem, pos):
            if form == word:
                ans.append(stem + "+" + pos + "+" + feature)

    ans = sorted(set(ans))

    if ans:
        print(", ".join(ans))
    else:
        print("UNKNOWN")w
```

```python
vowels = "aeiou"

def con(w, i):
    if w[i] in vowels:
        return False
    if w[i] == "y" and i > 0:
        return not con(w, i - 1)
    return True

def measure(w):
    m = 0
    prev = False
    for i in range(len(w)):
        c = not con(w, i)
        if c and not prev:
            m += 1
        prev = c
    return m

def vowel(w):
    for i in range(len(w)):
        if not con(w, i):
            return True
    return False

def double(w):
    return len(w) >= 2 and con(w, len(w)-1) and w[-1] == w[-2]

def short_o(w):
    if len(w) < 3:
        return False
    return con(w, -1) and not con(w, -2) and con(w, -3) and w[-1] not in "wxy"
def stem(w):
    if w.endswith("sses"):
        w = w[:-2]
    elif w.endswith("ies"):
        w = w[:-2]
    elif w.endswith("ss"):
        pass
    elif w.endswith("s"):
        w = w[:-1]
    if w.endswith("eed"):
        x = w[:-3]
        if measure(x) > 0:
            w = x + "ee"
    elif w.endswith("ed") and vowel(w[:-2]):
        w = w[:-2]
        if w.endswith(("at", "bl", "iz")):
            w += "e"
        elif double(w) and w[-1] not in "lsz":
            w = w[:-1]
        elif measure(w) == 1 and short_o(w):
            w += "e"
    elif w.endswith("ing") and vowel(w[:-3]):
        w = w[:-3]
        if w.endswith(("at", "bl", "iz")):
            w += "e"
        elif double(w) and w[-1] not in "lsz":
            w = w[:-1]
        elif measure(w) == 1 and short_o(w):
            w += "e"
    if w.endswith("y") and vowel(w[:-1]):
        w = w[:-1] + "i"
    return w
q = int(input())
for i in range(q):
    print(stem(input()))
```
