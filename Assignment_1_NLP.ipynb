```python
ASSIGNMENT 1

Task 1 PROGRAM

record = """
Patient had fractures in the left arm.
Paracetamol 500mg was given twice daily.
Patient was recovering well.
"""

levels = {
    "Phonology": "Study of speech sounds",
    "Morphology": "Study of word formation",
    "Syntax": "Study of sentence structure",
    "Semantics": "Study of word meaning",
    "Pragmatics": "Meaning based on context",
    "Discourse": "Study of connected sentences"
}

print("LEVELS OF LINGUISTIC ANALYSIS")

for level, meaning in levels.items():
    print(level, ":", meaning)

models = {
    "State Machine": "Validates patient IDs and dates",
    "Rule Based Model": "Recognizes clinical patterns",
    "Probabilistic Model": "Handles uncertain or misspelled words"
}

print("\nMODEL FAMILIES")

for model, use in models.items():
    print(model, ":", use)

pipeline = [
    "Input Clinical Records",
    "Text Tokenization",
    "Regular Expression Matching",
    "Finite State Automata",
    "Morphological Analysis",
    "Porter Stemming",
    "Search Results"
]

print("\nNLP PROCESSING PIPELINE")

for step in pipeline:
    print(step)


Task 2 PROGRAM

import re

text = """
Patient ID: P12345
Admission Date: 12/03/2025
Discharge Date: 15/03/2025
Drug: Paracetamol 500mg
Frequency: Twice daily
Lab: Hb 12.5 g/dL
BP: 120/80 mmHg
Temperature: 37.5 C
ICD Code: N18.9
Ward A bed 12
Contact: 9876543210
Age: 45 years
Symptom Duration: 3 days
"""

patterns = {
    "Patient ID": r"P\d{5}",
    "Admission Date": r"\d{2}/\d{2}/\d{4}",
    "Discharge Date": r"\d{2}/\d{2}/\d{4}",
    "Drug Strength": r"[A-Za-z]+\s*\d+\s*mg",
    "Dosage Frequency": r"Twice daily",
    "Lab Value": r"Hb\s*\d+\.\d+\s*g/dL",
    "Blood Pressure": r"\d{2,3}/\d{2,3}\s*mmHg",
    "Temperature": r"\d+\.\d+\s*C",
    "ICD Code": r"[A-Z]\d{2}\.\d",
    "Ward and Bed": r"Ward\s+[A-Z]\s+bed\s+\d+",
    "Contact Number": r"\b\d{10}\b",
    "Age": r"\d+\s+years",
    "Symptom Duration": r"Symptom Duration:\s*\d+\s+days"
}

print("HOSPITAL RECORD INFORMATION")

for name, pattern in patterns.items():
    result = re.findall(pattern, text, re.IGNORECASE)

    print(name, ":", result)


Task 3 PROGRAM
import re

text = """
Patient has HTN and DM.
BP is 120/80 mmHg.
Give Paracetamol 500mg OD.
Patient has fever for 3 days.
"""

tokens = re.findall(
    r'\d+(?:\.\d+)?\s*(?:mg|mmHg)|'
    r'[A-Za-z]+(?:-[A-Za-z]+)*|'
    r'\d+(?:\.\d+)?',
    text
)

print("ORIGINAL TEXT:")
print(text)

print("\nTOKENS:")
print(tokens)

abbreviations = {
    "HTN": "Hypertension",
    "DM": "Diabetes Mellitus",
    "BP": "Blood Pressure",
    "OD": "Once Daily",
    "BD": "Twice Daily",
    "TID": "Three Times Daily",
    "QID": "Four Times Daily",
    "Tab": "Tablet",
    "Cap": "Capsule",
    "Inj": "Injection",
    "IV": "Intravenous",
    "IM": "Intramuscular",
    "PO": "By Mouth",
    "PRN": "As Needed",
    "STAT": "Immediately",
    "Temp": "Temperature",
    "HR": "Heart Rate",
    "CBC": "Complete Blood Count",
    "Hb": "Haemoglobin",
    "WBC": "White Blood Cells",
    "RBC": "Red Blood Cells",
    "ECG": "Electrocardiogram",
    "CT": "Computed Tomography",
    "MRI": "Magnetic Resonance Imaging",
    "ICU": "Intensive Care Unit",
    "HIV": "Human Immunodeficiency Virus",
    "UTI": "Urinary Tract Infection",
    "COPD": "Chronic Obstructive Pulmonary Disease",
    "CNS": "Central Nervous System",
    "ENT": "Ear Nose and Throat"
}

normalized_text = text

for short, full in abbreviations.items():
    normalized_text = re.sub(
        r'\b' + short + r'\b',
        full,
        normalized_text,
        flags=re.IGNORECASE
    )

print("\nNORMALIZED TEXT:")
print(normalized_text)




print("1. DFA IMPLEMENTATION")
print("---------------------")

dfa_transitions = {
    "q0": {"a": "q1", "b": "q0"},
    "q1": {"a": "q1", "b": "q2"},
    "q2": {"a": "q1", "b": "q0"}
}

def dfa_accept(word):
    current_state = "q0"

    for symbol in word:
        if symbol not in ["a", "b"]:


Task 5 PROGRAM
print("1. INFLECTIONAL MORPHOLOGY")
print("--------------------------")

words = [
    "play", "plays", "played", "playing",
    "cat", "cats",
    "walk", "walked", "walking",
    "book", "books"
]

for word in words:

    if word.endswith("ing"):
        print(word, "-> Present participle")

    elif word.endswith("ed"):
        print(word, "-> Past tense")

    elif word.endswith("s"):
        print(word, "-> Plural / Third-person singular")

    else:
        print(word, "-> Base form")


print("\n2. INFLECTIONAL PARADIGMS")
print("-------------------------")

paradigms = {
    "Play (Verb)": ["play", "plays", "played", "playing"],
    "Walk (Verb)": ["walk", "walks", "walked", "walking"],
    "Cat (Noun)": ["cat", "cats"],
    "Book (Noun)": ["book", "books"]
}

for root, forms in paradigms.items():

    print(root, ":", ", ".join(forms))


print("\n3. ORTHOGRAPHIC RULES")
print("---------------------")

def plural_form(word):

    if word.endswith("y") and word[-2] not in "aeiou":
        return word[:-1] + "ies"

    elif word.endswith(("s", "x", "ch", "sh")):
        return word + "es"

    else:
        return word + "s"

def past_tense(word):

    if word.endswith("e"):
        return word + "d"

    elif word.endswith("y") and word[-2] not in "aeiou":
        return word[:-1] + "ied"

    else:
        return word + "ed"

def present_participle(word):

    if word.endswith("ie"):
        return word[:-2] + "ying"

    elif word.endswith("e") and not word.endswith("ee"):
        return word[:-1] + "ing"

    else:
        return word + "ing"

print("Plural of city:", plural_form("city"))
print("Plural of watch:", plural_form("watch"))
print("Past tense of study:", past_tense("study"))
print("Past tense of play:", past_tense("play"))
print("Present participle of make:", present_participle("make"))
print("Present participle of lie:", present_participle("lie"))


print("\n4. IRREGULAR INFLECTIONAL FORMS")
print("--------------------------------")

irregular_words = {
    "went": "go",
    "gone": "go",
    "children": "child",
    "mice": "mouse",
    "feet": "foot",
    "better": "good",
    "worse": "bad",
    "men": "man",
    "women": "woman"
}

for word, base in irregular_words.items():

    print(word, "->", base)


print("\n5. TYPE-TO-LEMMA RATIO")
print("----------------------")

text = """
Patients were treated.
The patient was taking medicines.
Children were playing.
The patients received medicine.
"""

tokens = [
    word.strip(".,!?").lower()
    for word in text.split()
]


lemma_dictionary = {
    "patients": "patient",
    "were": "be",
    "treated": "treat",
    "was": "be",
    "taking": "take",
    "medicines": "medicine",
    "children": "child",
    "playing": "play",
    "received": "receive"
}

lemmas = []

for word in tokens:

    lemma = lemma_dictionary.get(word, word)
    lemmas.append(lemma)

unique_types = set(tokens)
unique_lemmas = set(lemmas)

type_count = len(unique_types)
lemma_count = len(unique_lemmas)

ratio = type_count / lemma_count

print("Total Tokens:", len(tokens))
print("Unique Types:", type_count)
print("Unique Lemmas:", lemma_count)
print("Type-to-Lemma Ratio:", round(ratio, 2))

print("\nUnique Lemmas:")
print(", ".join(sorted(unique_lemmas)))


print("\n6. FINAL SUMMARY")
print("----------------")

print("Inflectional forms identified successfully.")
print("Inflectional paradigms generated.")
print("Orthographic rules demonstrated.")
print("Irregular forms mapped to their base words.")
print("Type-to-lemma ratio calculated successfully.")


Task 6 PROGRAM
print("1. DERIVATIONAL AFFIX IDENTIFICATION")
print("------------------------------------")

prefixes = [
    "un", "re", "dis", "mis", "pre",
    "non", "anti"
]

suffixes = [
    "ness", "ment", "tion", "er",
    "ful", "less", "able", "ly"
]

words = [
    "unhappy", "rewrite", "disagree",
    "misunderstand", "preview", "nonmedical",
    "antibiotic", "kindness", "development",
    "medication", "teacher", "careful",
    "careless", "readable", "quickly"
]

print("Prefix and Suffix Analysis:\n")

for word in words:

    found = []

    for prefix in prefixes:
        if word.startswith(prefix):
            found.append("Prefix: " + prefix)

    for suffix in suffixes:
        if word.endswith(suffix):
            found.append("Suffix: " + suffix)

    if found:
        print(word, "->", ", ".join(found))
    else:
        print(word, "-> No listed affix")


print("\n2. DERIVATIONAL WORD FAMILIES")
print("------------------------------")

word_families = {
    "happy": ["unhappy", "happiness", "happily"],
    "care": ["careful", "careless", "carefully"],
    "teach": ["teacher", "teaching", "teachable"],
    "develop": ["development", "developer"],
    "act": ["action", "active", "activity", "actor"],
    "medicine": ["medicinal", "medication"]
}

for root, derived_words in word_families.items():

    print("\nRoot Word:", root)

    for word in derived_words:
        print("-", word)


print("\n3. SEMANTIC ANALYSIS")
print("--------------------")

semantic_examples = {
    "unhappy": "Not happy",
    "careless": "Without care",
    "teacher": "A person who teaches",
    "inflammable": "Means easily catches fire; not the opposite of flammable",
    "disgruntled": "Means dissatisfied or annoyed; its meaning is not easily derived from the
parts"
}

for word, meaning in semantic_examples.items():

    print(word, ":", meaning)


print("\n4. AFFIX PRODUCTIVITY ANALYSIS")
print("------------------------------")

sample_words = [
    "kindness", "happiness", "sadness",
    "development", "movement", "treatment",
    "teacher", "writer", "worker",
    "careful", "hopeful", "useful",
    "careless", "fearless", "harmless"
]

affixes = [
    "ness", "ment", "er", "ful", "less"
]

productivity_count = {}

for affix in affixes:

    count = 0

    for word in sample_words:
        if word.endswith(affix):
            count += 1

    productivity_count[affix] = count

for affix, count in productivity_count.items():

    print(
        "-" + affix,
        ":", count,
        "words found"
    )
most_productive = max(
    productivity_count,
    key=productivity_count.get
)

print("\nMost frequent affix in sample:", "-" + most_productive)

print("\n5. MORPHOLOGICAL CONFLATION ANALYSIS")
print("------------------------------------")

word_groups = {
    "medic": ["medicine", "medicinal", "medication"],
    "care": ["care", "careful", "careless"],
    "act": ["act", "action", "active", "activity"]
}

for root, related_words in word_groups.items():

    print("\nConflated Group:", root)
    print("Words:", ", ".join(related_words))


print("\nPossible Incorrect Conflation:")

print("patient -> person receiving medical treatment")
print("patience -> ability to tolerate delay")

print("These words look similar but have different meanings.")


print("\n6. FINAL SUMMARY")
print("----------------")

print("15 derivational affixes identified.")
print("Derivational word families generated.")
print("Semantic changes and exceptions analysed.")
print("Affix productivity estimated from sample words.")
print("Morphological conflation demonstrated.")


Task 7 PROGRAM (PDF pages 23–26)
---------------------------------------------
print("1. MORPHOLOGICAL LEXICON")
print("-----------------------")


lexicon = {
    "patient": ["NOUN"],
    "doctor": ["NOUN"],
    "medicine": ["NOUN"],
    "record": ["NOUN", "VERB"],
    "treat": ["VERB"],
    "monitor": ["VERB"]
}

for stem, categories in lexicon.items():
    print(stem, ":", ", ".join(categories))


print("\n2. CONTINUATION CLASSES")
print("-----------------------")

continuation_classes = {
    "NOUN": {
        "s": "PLURAL"
    },

    "VERB": {
        "s": "PRESENT_3SG",
        "ed": "PAST",
        "ing": "PRESENT_PARTICIPLE"
    }
}

for category, suffixes in continuation_classes.items():

    print(category, "->", suffixes)


print("\n3. FINITE-STATE TRANSDUCER")
print("--------------------------")

def morphological_analysis(word):

    analyses = []

    for stem, categories in lexicon.items():

        if not word.startswith(stem):
            continue

        suffix = word[len(stem):]

        for category in categories:

            if suffix == "":
                analyses.append(
                    stem + "+" + category
                )

            elif suffix in continuation_classes[category]:

                grammatical_feature = (
                    continuation_classes[category][suffix]
                )

                analyses.append(
                    stem + "+" + category +
                    "+" + grammatical_feature
                )

    return analyses


print("\n4. MORPHOLOGICAL ANALYSIS")
print("------------------------")

test_words = [
    "patients",
    "doctors",
    "medicines",
    "treated",
    "treating",
    "records",
    "monitoring"
]

for word in test_words:

    result = morphological_analysis(word)

    print("\nWord:", word)

    if result:
        for analysis in result:
            print("Analysis:", analysis)
    else:
        print("No analysis found")


print("\n5. MORPHOLOGICAL GENERATION")
print("---------------------------")

def generate_word(stem, category, feature):

    if stem not in lexicon:
        return "Stem not found"

    if category not in lexicon[stem]:
        return "Invalid grammatical category"

    if feature == "BASE":
        return stem

    for suffix, grammatical_feature in (
        continuation_classes[category].items()
    ):

        if grammatical_feature == feature:
            return stem + suffix

    return "Invalid grammatical feature"

generation_examples = [
    ("patient", "NOUN", "PLURAL"),
    ("doctor", "NOUN", "PLURAL"),
    ("treat", "VERB", "PAST"),
    ("treat", "VERB", "PRESENT_PARTICIPLE"),
    ("record", "VERB", "PRESENT_3SG")
]

for stem, category, feature in generation_examples:

    generated = generate_word(
        stem, category, feature
    )

    print(
        stem + "+" + category + "+" + feature,
        "->",
        generated
    )


print("\n6. MORPHOLOGICAL AMBIGUITY")
print("--------------------------")

ambiguous_word = "records"

analyses = morphological_analysis(ambiguous_word)

print("Word:", ambiguous_word)

for analysis in analyses:
    print("Possible Analysis:", analysis)

if len(analyses) > 1:
    print("Result: Morphologically Ambiguous")
else:
    print("Result: Unambiguous")


print("\n7. FINAL SUMMARY")
print("----------------")

print("Lexicon created successfully.")
print("Continuation classes defined.")
print("Finite-State Transducer implemented.")
print("Word forms analysed successfully.")
print("New word forms generated successfully.")
print("Morphological ambiguity identified.")


Task 9 PROGRAM
import nltk
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()


print("1. PORTER STEMMING")
print("------------------")

words = [
    "caresses",
    "ponies",
    "cats",
    "running",
    "hopping",
    "happy",
    "relational",
    "conditional",
    "digitizer",
    "activation"
]

for word in words:
    print(word, "->", stemmer.stem(word))


print("\n2. PORTER STEMMER CONDITIONS")
print("----------------------------")


def cv_sequence(word):

    result = ""

    for i, char in enumerate(word):

        if char in "aeiou":
            result += "V"

        elif char == "y" and i > 0 and word[i - 1] in "aeiou":
            result += "V"

        else:
            result += "C"

    return result


def calculate_m(word):

    sequence = cv_sequence(word)

    count = 0

    for i in range(len(sequence) - 1):

        if sequence[i:i + 2] == "VC":
            count += 1

    return count


def contains_vowel(word):
    return any(char in "aeiou" for char in word)


def double_consonant(word):

    return (
        len(word) >= 2 and
        word[-1] == word[-2] and
        word[-1] not in "aeiou"
    )


def cvc_condition(word):

    if len(word) < 3:
        return False

    sequence = cv_sequence(word)

    return (
        sequence[-3:] == "CVC" and
        word[-1] not in "wxy"
    )

test_words = ["tr", "trouble", "tree", "running", "hop"]

for word in test_words:

    print("\nWord:", word)
    print("CV Pattern:", cv_sequence(word))
    print("Measure m:", calculate_m(word))
    print("Contains vowel (*v):", contains_vowel(word))
    print("Double consonant (*d):", double_consonant(word))
    print("CVC condition (*o):", cvc_condition(word))


print("\n3. FIVE MAJOR PORTER STEMMING STEPS")
print("-----------------------------------")


step_examples = {
    "Step 1 - Plural and tense handling": [
        ("caresses", "caress"),
        ("ponies", "poni"),
        ("cats", "cat"),
        ("running", "run"),
        ("hopping", "hop")
    ],

    "Step 2 - Derivational suffix replacement": [
        ("relational", "relate"),
        ("conditional", "condition"),
        ("digitizer", "digitize")
    ],

    "Step 3 - Further suffix reduction": [
        ("triplicate", "triplic"),
        ("formative", "form"),
        ("formalize", "formal")
    ],

    "Step 4 - Suffix removal": [
        ("revival", "reviv"),
        ("allowance", "allow"),
        ("adjustment", "adjust")
    ],

    "Step 5 - Final e and l removal": [
        ("probate", "probat"),
        ("rate", "rate"),
        ("controll", "control")
    ]
}

for step, examples in step_examples.items():

    print("\n" + step)

    for original, result in examples:
        print(original, "->", result)


print("\n4. STEMMING VS LEMMATISATION")
print("----------------------------")


lemma_dictionary = {
    "running": "run",
    "ran": "run",
    "better": "good",
    "mice": "mouse",
    "children": "child",
    "studies": "study",
    "cars": "car"
}

comparison_words = [
    "running",
    "ran",
    "better",
    "mice",
    "children",
    "studies",
    "cars"
]

for word in comparison_words:

    stem = stemmer.stem(word)

    lemma = lemma_dictionary.get(word, word)

    print("\nWord:", word)
    print("Stem:", stem)
    print("Lemma:", lemma)


print("\n5. STEMMING ERRORS")
print("------------------")

error_examples = [
    ("university", "univers", "Over-stemming"),
    ("organization", "organ", "Possible over-stemming"),
    ("policy", "polici", "Under-stemming"),
    ("running", "run", "Correct stemming")
]

for word, stem, error_type in error_examples:

    print(word, "->", stem, ":", error_type)


print("\n6. INFORMATION RETRIEVAL EVALUATION")
print("-----------------------------------")

retrieved_documents = {"D1", "D2", "D3", "D4"}

relevant_documents = {"D1", "D2", "D5"}


true_positive = len(
    retrieved_documents.intersection(relevant_documents)
)

false_positive = len(
    retrieved_documents - relevant_documents
)

false_negative = len(
    relevant_documents - retrieved_documents
)

precision = true_positive / (
    true_positive + false_positive
)

recall = true_positive / (
    true_positive + false_negative
)

f1_score = (
    2 * precision * recall /
    (precision + recall)
)

print("True Positives:", true_positive)
print("False Positives:", false_positive)
print("False Negatives:", false_negative)

print("Precision:", round(precision, 2))
print("Recall:", round(recall, 2))
print("F1-Score:", round(f1_score, 2))


print("\n7. FINAL SUMMARY")
print("----------------")

print("Porter Stemmer implemented successfully.")
print("Five major stemming stages demonstrated.")
print("Morphological conditions calculated.")
print("Stemming and lemmatisation compared.")
print("Stemming errors demonstrated.")
print("Information retrieval metrics calculated.")


Task 10 PROGRAM
import re
import time
import matplotlib.pyplot as plt
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()


print("1. SAMPLE HOSPITAL RECORDS")
print("--------------------------")

records = [
    "Pt has fever and SOB on 12/11/2026.",
    "BP 120/80 mmHg and temperature 38 C.",
    "Give 500 mg medicine to the patient.",
    "Dr reviewed the patient on 13/11/2026.",
    "Patient has DM and HTN.",
    "The patient received 250 mg medicine."
]

for i, record in enumerate(records, start=1):
    print("Record", i, ":", record)


print("\n2. ABBREVIATION EXPANSION")
print("-------------------------")

abbreviations = {
    "pt": "patient",
    "dr": "doctor",
    "sob": "shortness of breath",
    "bp": "blood pressure",
    "dm": "diabetes mellitus",
    "htn": "hypertension"
}

def expand_abbreviations(text):

    for short_form, full_form in abbreviations.items():

        pattern = r"\b" + re.escape(short_form) + r"\b"

        text = re.sub(
            pattern,
            full_form,
            text,
            flags=re.IGNORECASE
        )

    return text


print("\n3. DATE NORMALISATION")
print("--------------------")

def normalize_dates(text):


    def replace_date(match):

        day = match.group(1)
        month = match.group(2)
        year = match.group(3)

        return year + "-" + month.zfill(2) + "-" + day.zfill(2)

    return re.sub(
        r"\b(\d{1,2})/(\d{1,2})/(\d{4})\b",
        replace_date,
        text
    )


print("\n4. MEDICAL UNIT NORMALISATION")
print("-----------------------------")

def normalize_units(text):

    return re.sub(
        r"(\d+(?:\.\d+)?)\s*(mg|g|ml|mmhg)\b",
        lambda match: match.group(1) + match.group(2),
        text,
        flags=re.IGNORECASE
    )


print("\n5. TOKENISATION")
print("---------------")

def tokenize(text):

    pattern = (
        r"\d{4}-\d{2}-\d{2}|"
        r"\d{2,3}/\d{2,3}(?:mmhg)?|"
        r"\d+(?:\.\d+)?(?:mg|ml|mmhg|g)?|"
        r"[a-z]+(?:-[a-z]+)*"
    )

    return re.findall(pattern, text.lower())


print("\n6. COMPLETE NLP PIPELINE")
print("------------------------")

def nlp_pipeline(text):

    text = text.lower()

    text = expand_abbreviations(text)

    text = normalize_dates(text)

    text = normalize_units(text)

    tokens = tokenize(text)

    stems = [stemmer.stem(word) for word in tokens]

    return text, tokens, stems


start_time = time.perf_counter()

processed_records = []

for record in records:

    normalized, tokens, stems = nlp_pipeline(record)

    processed_records.append({
        "normalized": normalized,
        "tokens": tokens,
        "stems": stems
    })

end_time = time.perf_counter()

execution_time = end_time - start_time


print("Original Record:")
print(records[0])

print("\nNormalized Record:")
print(processed_records[0]["normalized"])

print("\nTokens:")
print(processed_records[0]["tokens"])

print("\nStemmed Words:")
print(processed_records[0]["stems"])


print("\n7. TEST CASE EVALUATION")
print("-----------------------")

test_cases = [
    ("Pt has fever.", "patient has fever."),
    ("BP 120/80 mmHg", "blood pressure 120/80mmhg"),
    ("Review on 12/11/2026", "review on 2026-11-12"),
    ("Give 500 mg", "give 500mg"),
    ("Dr treated the patient.", "doctor treated the patient.")
]

passed = 0

for input_text, expected in test_cases:

    actual, tokens, stems = nlp_pipeline(input_text)

    if actual == expected:
        result = "PASS"
        passed += 1
    else:
        result = "FAIL"

    print("\nInput:", input_text)
    print("Expected:", expected)
    print("Actual:", actual)
    print("Result:", result)

total_tests = len(test_cases)
failed = total_tests - passed

accuracy = (passed / total_tests) * 100

print("\nTotal Test Cases:", total_tests)
print("Passed:", passed)
print("Failed:", failed)
print("Test Accuracy:", round(accuracy, 2), "%")


print("\n8. EXECUTION PERFORMANCE")
print("-------------------------")

total_tokens = sum(
    len(record["tokens"])
    for record in processed_records
)

print("Records Processed:", len(processed_records))
print("Total Tokens:", total_tokens)
print("Execution Time:", round(execution_time, 6), "seconds")


print("\n9. PERFORMANCE VISUALISATION")
print("-----------------------------")

labels = ["Passed Tests", "Failed Tests"]
values = [passed, failed]

plt.figure(figsize=(7, 5))

plt.bar(labels, values)

plt.title("NLP Pipeline Test Results")
plt.xlabel("Test Cases")
plt.ylabel("Number of Tests")

plt.show()


print("\n10. FINAL SUMMARY")
print("-----------------")

print("Abbreviation expansion completed.")
print("Date normalisation completed.")
print("Medical unit normalisation completed.")
print("Tokenisation completed.")
print("Porter stemming completed.")
print("Test cases evaluated.")
print("Execution time measured.")
print("Performance graph generated.")


Task 11 PROGRAM
from nltk.stem import PorterStemmer
import re

stemmer = PorterStemmer()

documents = {
    "D1": "Patient treated with medicine",
    "D2": "Patients are treated",
    "D3": "Patient treatment record",
    "D4": "Doctor treated patient",
    "D5": "Medicines prescribed to patient"
}

query = "patients treated"

relevant_docs = {"D1", "D2", "D4"}

def tokenize(text):
    return re.findall(r"[a-z]+", text.lower())

def exact_normalize(word):
    return word

def porter_normalize(word):
    return stemmer.stem(word)

lemma_dict = {
    "patients": "patient",
    "patient": "patient",
    "treated": "treat",
    "treating": "treat",
    "treats": "treat",
    "medicines": "medicine"
}

def morphology_normalize(word):
    return lemma_dict.get(word, word)

def search_documents(normalizer):
    query_words = {
        normalizer(word) for word in tokenize(query)
    }

    results = []

    for doc_id, text in documents.items():
        doc_words = {
            normalizer(word) for word in tokenize(text)
        }

        if query_words.issubset(doc_words):
            results.append(doc_id)

    return set(results)

def evaluate(results):
    tp = len(results & relevant_docs)
    fp = len(results - relevant_docs)
    fn = len(relevant_docs - results)

    precision = tp / (tp + fp) if tp + fp else 0
    recall = tp / (tp + fn) if tp + fn else 0

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0
    )

    return precision, recall, f1

methods = {
    "Exact Matching": exact_normalize,
    "Porter Stemming": porter_normalize,
    "Morphology-Aware": morphology_normalize
}

print("NORMALISATION COMPARISON")
print("-" * 65)

for method_name, normalizer in methods.items():
    results = search_documents(normalizer)
    precision, recall, f1 = evaluate(results)

    print("\nMethod:", method_name)
    print("Retrieved Documents:", sorted(results))
    print("Precision:", round(precision, 2))
    print("Recall:", round(recall, 2))
    print("F1 Score:", round(f1, 2))

print("\nRECOMMENDATION")
print("Use morphology-aware normalisation with")
print("clinical abbreviation and terminology rules.")
print("It improves matching while preserving medical meaning.")
```
