```python
word = input("Enter a word: ")
if word.endswith("ed"):
    root = word[:-2]
    suffix = "ed"
    type = "Inflectional"
elif word.endswith("ing"):
    root = word[:-3]
    suffix = "ing"
    type = "Inflectional"
elif word.endswith("ion"):
    root = word[:-3]
    suffix = "ion"
    type = "Derivational"
else:
    root = word
    suffix = "-"
    type = "None"
print("Word:", word)
print("Root:", root)
print("Suffix:", suffix)
print("Type:", type)
print("Normalized form:", root)
```

```python
word = input("Enter a word: ")
if word.startswith("un"):
    prefix = "un"
    root = word[2:]
    suffix = "-"
elif word.endswith("ness"):
    prefix = "-"
    root = word[:-4]
    suffix = "ness"
elif word.endswith("ly"):
    prefix = "-"
    root = word[:-2]
    suffix = "ly"
else:
    prefix = "-"
    root = word
    suffix = "-"
print("Word:", word)
print("Prefix:", prefix)
print("Root:", root)
print("Suffix:", suffix)
print("Normalized root:", root)
```

```python
word = input("Enter a word: ")

if word == "writes":
    root = "write"
    states = "START -> write -> s -> END"
    type = "Regular"
elif word == "writing":
    root = "write"
    states = "START -> write -> ing -> END"
    type = "Regular"
elif word == "written":
    root = "write"
    states = "START -> written -> END"
    type = "Irregular"
else:
    root = word
    states = "START -> word -> END"
    type = "Unknown"

print("Word:", word)
print("State Path:", states)
print("Root:", root)
print("Type:", type)
```
