```python
import re

text = input("Enter a sentence: ")
pattern = input("Enter a word or pattern to search: ")

matches = re.findall(pattern, text, re.IGNORECASE)

if matches:
    print("Matches:", matches)
else:
    print("No match found")

print("Does the text contain the pattern?", bool(re.search(pattern, text, re.IGNORECASE)))
```
