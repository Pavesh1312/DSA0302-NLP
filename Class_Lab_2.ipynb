```python
def accepts(text):
    state = 0

    for ch in text:
        if state == 0:
            if ch == "a":
                state = 1
            else:
                state = 0
        elif state == 1:
            if ch == "a":
                state = 1
            elif ch == "b":
                state = 2
            else:
                state = 0
        else:
            if ch == "a":
                state = 1
            else:
                state = 0

    return state == 2

text = input("Enter a string containing a and b: ")

if accepts(text):
    print("Accepted")
else:
    print("Rejected")
```
