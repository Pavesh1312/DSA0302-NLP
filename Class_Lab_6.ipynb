```python
import random
from collections import defaultdict

text = input("Enter training text: ").lower().split()
bigrams = defaultdict(list)

for i in range(len(text) - 1):
    bigrams[text[i]].append(text[i + 1])

word = input("Enter the starting word: ").lower()
result = [word]

for i in range(10):
    if word not in bigrams:
        break
    word = random.choice(bigrams[word])
    result.append(word)

print("Generated text:", " ".join(result))
```
