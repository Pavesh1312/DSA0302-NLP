```python
import nltk

patterns = [
    (r".*ing$", "VBG"),
    (r".*ed$", "VBD"),
    (r".*ly$", "RB"),
    (r".*ous$", "JJ"),
    (r".*s$", "NNS"),
    (r".*", "NN")
]

tagger = nltk.RegexpTagger(patterns)
sentence = input("Enter a sentence: ")
words = nltk.word_tokenize(sentence)
print(tagger.tag(words))
```
