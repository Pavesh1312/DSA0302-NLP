```python
words = ["analyzing", "analysis", "analytical"]
for word in words:
    if word == "analyzing":
        root = "analyze"
        affix = "ing"
        form = "Inflectional"
    elif word == "analysis":
        root = "analy"
        affix = "sis"
        form = "Derivational"
    else:
        root = "analyze"
        affix = "ical"
        form = "Derivational"
    print("Word:", word)
    print("Root:", root)
    print("Affix:", affix)
    print("Type:", form)
    print("Normalized:", "analyze")
    print()
```

```python
words = ["disagree", "agreement", "agreeable"]
for word in words:
    if word == "disagree":
        prefix = "dis"
        root = "agree"
        suffix = "-"
        meaning = "opposite of agree"
    elif word == "agreement":
        prefix = "-"
        root = "agree"
        suffix = "ment"
        meaning = "state of agreeing"
    else:
        prefix = "-"
        root = "agree"
        suffix = "able"
        meaning = "capable of agreeing"
    print("Word:", word)
    print("Prefix:", prefix)
    print("Root:", root)
    print("Suffix:", suffix)
    print("Meaning:", meaning)
    print("Normalized:", "agree")
    print()
```

```python
words = ["govern", "government", "governance"]
for word in words:
    if word == "govern":
        root = "govern"
        affix = "-"
        level = "Base"
    elif word == "government":
        root = "govern"
        affix = "ment"
        level = "Derivational"
    else:
        root = "govern"
        affix = "ance"
        level = "Derivational"
    print("Word:", word)
    print("Root:", root)
    print("Affix:", affix)
    print("Level:", level)
    print("Normalized:", "govern")
    print()
```

```python
words = ["activate", "activation", "reactivation"]
for word in words:
    if word == "activate":
        prefix = "-"
        root = "activate"
        suffix = "-"
        sequence = "activate"
    elif word == "activation":
        prefix = "-"
        root = "activate"
        suffix = "ion"
        sequence = "activate -> activation"
    else:
        prefix = "re"
        root = "activate"
        suffix = "ion"
        sequence = "activate -> reactivate -> reactivation"
    print("Word:", word)
    print("Prefix:", prefix)
    print("Root:", root)
    print("Suffix:", suffix)
    print("Sequence:", sequence)
    print("Normalized:", "activate")
    print()
```

```python
words = ["create", "creates", "creating"]
for word in words:
    if word == "create":
        root = "create"
        suffix = "-"
        category = "Base Form"
    elif word == "creates":
        root = "create"
        suffix = "s"
        category = "Third Person Singular"
    else:
        root = "create"
        suffix = "ing"
        category = "Present Participle"
    print("Word:", word)
    print("Root:", root)
    print("Suffix:", suffix)
    print("Category:", category)
    print("Normalized:", "create")
    print()
```
