```python
def make_plural(word):
    if word.endswith(("s", "sh", "ch", "x", "z")):
        return word + "es"
    elif word.endswith("y") and len(word) > 1 and word[-2] not in "aeiou":
        return word[:-1] + "ies"
    else:
        return word + "s"

word = input("Enter a singular noun: ")
print("Plural form:", make_plural(word))
```
